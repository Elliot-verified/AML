```python
# pipeline.py

from __future__ import annotations

from dataclasses import dataclass, asdict
from pathlib import Path
import json
from typing import Dict, List

import pandas as pd
import matplotlib.pyplot as plt

from Bio import SeqIO, pairwise2

from PepMLM_gen_scripts.pepmlm_generate_fast import PepMLMConfig, run_pepmlm
from MetaLATTE_gen_scripts.run_metalatte_fasta import run_metalatte_on_fasta


# ============================================================
# CONFIG DATA STRUCTURES
# ============================================================

@dataclass
class SeedSet:
    name: str          # e.g. "pb_mer_operon"
    fasta_path: Path   # path to seeds fasta file


@dataclass
class PepMLMParams:
    n_variants_per_seed: int = 20
    fraction_mask: float = 0.25
    topk_per_mask: int = 3          
    max_len: int = 200
    batch_size: int = 8


@dataclass
class RunConfig:
    """
    A single experiment configuration (one 'run'):
      - same PepMLM params for all seed sets
      - novelty threshold, etc.
    """
    name: str
    out_dir: Path
    pepmlm_params: PepMLMParams
    novelty_identity_threshold: float = 0.95  # e.g. < 70% identity to be 'novel'
    top_k_variants: int = 200


# ============================================================
# UTILS
# ============================================================

def load_fasta_as_dict(fasta_path: Path) -> Dict[str, str]:
    """Return {id: sequence} from a FASTA file."""
    records = SeqIO.parse(str(fasta_path), "fasta")
    return {r.id: str(r.seq) for r in records}


from Bio.Align import PairwiseAligner

# Create one global aligner for speed
_global_aligner = PairwiseAligner()
_global_aligner.mode = "global"
_global_aligner.match_score = 1.0
_global_aligner.mismatch_score = 0.0
_global_aligner.open_gap_score = 0.0
_global_aligner.extend_gap_score = 0.0

def max_identity_to_any_seed(seq: str, seed_seqs: List[str]) -> float:
    """
    Compute max percent identity between seq and seed sequences
    using the modern Bio.Align.PairwiseAligner (replaces pairwise2).
    Equivalent to globalxx scoring (match=1, mismatch=0, no gap cost).
    """
    max_id = 0.0

    for seed in seed_seqs:
        # PairwiseAligner.align returns a generator of alignments
        alignments = _global_aligner.align(seq, seed)

        # Take the first (best) alignment
        aln = alignments[0]

        # Extract aligned sequences as strings with gaps
        aligned_query = aln.aligned[0]
        aligned_seed = aln.aligned[1]

        # Compute percent identity manually
        # (PairwiseAligner doesn't compute identity automatically)
        matches = aln.score  # since match=1, mismatch=0
        length = max(len(seq), len(seed))  # globalxx-like denominator

        ident = matches / length

        if ident > max_id:
            max_id = ident

    return max_id



def compute_binding_metrics(df: pd.DataFrame) -> pd.DataFrame:
    """
    Add binding_capacity and best_metal columns from MetaLATTE prob_* columns.
    Assumes columns like prob_Pb, prob_Hg, prob_Non_binding, etc.
    """
    metal_cols = [
        c for c in df.columns
        if c.startswith("prob_") and "Non" not in c and "non" not in c
    ]
    if not metal_cols:
        raise ValueError("No metal prob_ columns found in MetaLATTE output.")

    df["binding_capacity"] = df[metal_cols].max(axis=1)
    df["best_metal"] = df[metal_cols].idxmax(axis=1).str.replace("prob_", "")
    return df


# ============================================================
# CORE PIPELINE FOR ONE SEED SET WITHIN A RUN
# ============================================================

def run_for_seed_set(seed: SeedSet, run_cfg: RunConfig) -> pd.DataFrame:
    """
    Run PepMLM + MetaLATTE for a single seed set within a run,
    compute novelty and binding metrics, and return a dataframe.
    """
    run_dir = run_cfg.out_dir
    run_dir.mkdir(parents=True, exist_ok=True)

    seed_subdir = run_dir / f"seed_{seed.name}"
    seed_subdir.mkdir(exist_ok=True)

    # 1. Generate variants with PepMLM
    variants_fasta = seed_subdir / "generated_variants.fasta"

    pep_cfg = PepMLMConfig(
        fasta_in=seed.fasta_path,
        fasta_out=variants_fasta,
        n_variants_per_seed=run_cfg.pepmlm_params.n_variants_per_seed,
        fraction_mask=run_cfg.pepmlm_params.fraction_mask,
        topk_per_mask=run_cfg.pepmlm_params.topk_per_mask,
        max_len=run_cfg.pepmlm_params.max_len,
        batch_size=run_cfg.pepmlm_params.batch_size,
    )
    print(f"[{run_cfg.name}] Running PepMLM for seed set '{seed.name}'")
    run_pepmlm(pep_cfg)

    # 2. Run MetaLATTE on generated variants
    metalatte_csv = seed_subdir / "metalatte_predictions.csv"
    print(f"[{run_cfg.name}] Running MetaLATTE for seed set '{seed.name}'")
    run_metalatte_on_fasta(
        fasta_path=str(variants_fasta),
        out_csv_path=str(metalatte_csv),
    )

    df = pd.read_csv(metalatte_csv)

    # 3. Add binding metrics
    df = compute_binding_metrics(df)

    # 4. Compute novelty (max identity vs any seed sequence)
    seed_seqs = list(load_fasta_as_dict(seed.fasta_path).values())
    print(f"[{run_cfg.name}] Computing novelty vs seeds for '{seed.name}' "
          f"(threshold: {run_cfg.novelty_identity_threshold:.2f})")

    # assume MetaLATTE output has a 'sequence' column (adjust if needed)
    if "sequence" not in df.columns:
        raise ValueError(
            "Expected a 'sequence' column in MetaLATTE output; "
            "adjust code if it's named differently."
        )

    identities = []
    for seq in df["sequence"]:
        identities.append(max_identity_to_any_seed(seq, seed_seqs))
    df["max_identity_to_seed"] = identities
    df["is_novel"] = df["max_identity_to_seed"] < run_cfg.novelty_identity_threshold

    # annotate with seed set & run name
    df["seed_set"] = seed.name
    df["run_name"] = run_cfg.name

    # 5. Save per-seed-set top variants if desired
    top_df = df.sort_values("binding_capacity", ascending=False).head(run_cfg.top_k_variants)
    top_df.to_csv(seed_subdir / "top_variants.csv", index=False)

    # 6. Make a simple novelty / identity histogram plot
    plots_dir = run_dir / "plots"
    plots_dir.mkdir(exist_ok=True)
    plt.figure()
    plt.hist(df["max_identity_to_seed"], bins=30)
    plt.xlabel("Max sequence identity to any seed")
    plt.ylabel("Count of generated variants")
    plt.title(f"{run_cfg.name} – {seed.name} – similarity to seeds")
    plt.tight_layout()
    img_path = plots_dir / f"identity_hist_{seed.name}.png"
    plt.savefig(img_path, dpi=200, bbox_inches="tight")
    plt.close()
    print(f"[{run_cfg.name}] Saved similarity histogram → {img_path}")

    return df


# ============================================================
# RUNNING ONE EXPERIMENT (ALL SEED FILES AT ONCE)
# ============================================================

def run_experiment(seed_sets: List[SeedSet], run_cfg: RunConfig) -> None:
    """
    Run a full experiment:
      - all seed sets with the same PepMLM hyperparams,
      - aggregate into a single CSV,
      - compute per-seed summaries.
    """
    print("=" * 70)
    print(f"Running experiment: {run_cfg.name}")
    print("=" * 70)

    run_cfg.out_dir.mkdir(parents=True, exist_ok=True)

    # Save run config for reproducibility
    with open(run_cfg.out_dir / "run_config.json", "w") as f:
        json.dump(asdict(run_cfg), f, indent=2, default=str)

    all_variants = []

    for seed in seed_sets:
        df_seed = run_for_seed_set(seed, run_cfg)
        all_variants.append(df_seed)

    all_df = pd.concat(all_variants, ignore_index=True)

    # Save aggregated variants with scores for this run
    all_csv = run_cfg.out_dir / "all_variants_with_scores.csv"
    all_df.to_csv(all_csv, index=False)
    print(f"[{run_cfg.name}] Saved all variant scores → {all_csv}")

    # Per-seed summary: novelty percentage & binding capacity stats
    summary = (
        all_df
        .groupby("seed_set")
        .agg(
            n_variants=("sequence", "count"),
            novelty_pct=("is_novel", lambda x: 100.0 * x.mean()),
            mean_binding_capacity=("binding_capacity", "mean"),
            max_binding_capacity=("binding_capacity", "max"),
        )
        .reset_index()
    )

    summary_csv = run_cfg.out_dir / "summary_by_seed.csv"
    summary.to_csv(summary_csv, index=False)
    print(f"[{run_cfg.name}] Saved summary by seed set → {summary_csv}")

    print(f"[{run_cfg.name}] DONE.")


# ============================================================
# ABLATION ANALYSIS – COMPARING MULTIPLE RUNS
# ============================================================

def load_run_summary(run_dir: Path) -> pd.DataFrame:
    summary_path = run_dir / "summary_by_seed.csv"
    if not summary_path.exists():
        raise FileNotFoundError(f"Summary file not found: {summary_path}")
    df = pd.read_csv(summary_path)
    df["run_dir"] = str(run_dir)
    df["run_name"] = run_dir.name
    return df


def compare_runs(run_dirs: List[Path]) -> pd.DataFrame:
    """
    Load summaries from multiple runs and return combined dataframe.
    You can then plot novelty vs binding capacity for ablation analysis.
    """
    all_summaries = [load_run_summary(rd) for rd in run_dirs]
    combined = pd.concat(all_summaries, ignore_index=True)

    # Example quick plots (per run):
    plt.figure()
    for run_name, sub in combined.groupby("run_name"):
        plt.scatter(
            sub["novelty_pct"],
            sub["mean_binding_capacity"],
            label=run_name,
        )
    plt.xlabel("Novelty % (< threshold identity)")
    plt.ylabel("Mean binding capacity")
    plt.title("Run comparison: novelty vs binding capacity")
    plt.legend()
    plt.tight_layout()
    plt.show()

    return combined


# ============================================================
# EXAMPLE USAGE (EDIT FOR YOUR PROJECT)
# ============================================================

if __name__ == "__main__":
    # 1) Define all your seed FASTA files once.
    seed_sets = [
        SeedSet(
            name="pb_mer_operon",
            fasta_path=Path("PepMLM_seeds/Metal_specific/pb_seeds.fasta"),
        ),
        SeedSet(
            name="hg_specific",
            fasta_path=Path("PepMLM_seeds/Metal_specific/hg_seeds.fasta"),
        ),
        SeedSet(
            name="cd_specific",
            fasta_path=Path("PepMLM_seeds/Metal_specific/cd_seeds.fasta"),
        ),
        SeedSet(
            name="MT_bacterial_redox",
            fasta_path=Path("PepMLM_seeds/MT_seeds/MT_Bacterial_Redox.fasta"),
        ),
        SeedSet(
            name="MT_bacterial_transformation",
            fasta_path=Path("PepMLM_seeds/MT_seeds/MT_Bacterial_transformation.fasta"),
        ),
    ]

    # 2) Define different runs with different hyperparameters (for ablation)
    base_params = PepMLMParams(
        n_variants_per_seed=20,
        fraction_mask=0.25,   # fewer positions masked → more conservative
        topk_per_mask=3,      # or whatever you want to test
        max_len=200,
        batch_size=8,
    )

    runs = [
        RunConfig(
            name="run_base",
            out_dir=Path("notebook_runs/Final"),
            pepmlm_params=base_params,
            novelty_identity_threshold=0.95,
            top_k_variants=999999,
        ),
    ]

    # 3) Run all experiments (all seed sets for each hyperparameter configuration)
    for cfg in runs:
        run_experiment(seed_sets, cfg)

    # 4) After all runs complete, do ablation analysis
    combined_summary = compare_runs([r.out_dir for r in runs])
    print(combined_summary)
```

```python
# ============================================================
# TARGETED SELECTION BY METAL + NOVELTY
# ============================================================
from pathlib import Path
from typing import List
import pandas as pd
def select_variants_for_metals(
    run_dirs: List[Path],
    target_metals: List[str],
    novelty_threshold: float,
    min_binding_capacity: float = 0.0,
    top_k_per_run: int | None = None,
) -> pd.DataFrame:
    """
    From one or more runs, return variants that:
      - have best_metal in target_metals,
      - have max_identity_to_seed < novelty_threshold,
      - have binding_capacity >= min_binding_capacity (optional),
    and return their sequences, binding capacities, and similarity scores.

    Parameters
    ----------
    run_dirs : list of Path
        Directories that contain all_variants_with_scores.csv
        (e.g. [Path("notebook_runs/run_base")])
    target_metals : list of str
        Metals you're interested in, e.g. ["Fe"] or ["Fe", "Cu"].
        These must match the strings in df["best_metal"] (Cd, Cu, Fe, Hg, Mn, Ni, Zn, ...).
    novelty_threshold : float
        Maximum allowed identity to any seed (e.g. 0.95 for 95%).
    min_binding_capacity : float, optional
        Minimum predicted binding_capacity to keep (default 0.0).
    top_k_per_run : int or None, optional
        If set, keep only the top K variants per run (after filtering),
        sorted by binding_capacity descending.

    Returns
    -------
    pd.DataFrame
        Columns include at least:
          - sequence
          - binding_capacity
          - max_identity_to_seed
          - best_metal
          - seed_set
          - run_name
    """
    all_selected = []

    for run_dir in run_dirs:
        csv_path = run_dir / "all_variants_with_scores.csv"
        if not csv_path.exists():
            raise FileNotFoundError(f"Missing file: {csv_path}")

        df = pd.read_csv(csv_path)

        required_cols = {
            "sequence",
            "binding_capacity",
            "best_metal",
            "max_identity_to_seed",
            "seed_set",
            "run_name",
        }
        missing = required_cols - set(df.columns)
        if missing:
            raise ValueError(f"{csv_path} is missing required columns: {missing}")

        # Filter by metal, novelty, and binding capacity
        sub = df[
            (df["best_metal"].isin(target_metals)) &
            (df["max_identity_to_seed"] < novelty_threshold) &
            (df["binding_capacity"] >= min_binding_capacity)
        ].copy()

        # Optionally keep only top K by binding_capacity for this run
        if top_k_per_run is not None and len(sub) > 0:
            sub = sub.sort_values("binding_capacity", ascending=False).head(top_k_per_run)

        all_selected.append(sub)

    if not all_selected:
        return pd.DataFrame(
            columns=[
                "sequence",
                "binding_capacity",
                "max_identity_to_seed",
                "best_metal",
                "seed_set",
                "run_name",
            ]
        )

    result = pd.concat(all_selected, ignore_index=True)

    # Return the most relevant columns first
    cols_order = [
        "sequence",
        "binding_capacity",
        "max_identity_to_seed",
        "best_metal",
        "seed_set",
        "run_name",
    ]
    existing_cols_ordered = [c for c in cols_order if c in result.columns]
    result = result[existing_cols_ordered + [c for c in result.columns if c not in existing_cols_ordered]]

    return result
```
